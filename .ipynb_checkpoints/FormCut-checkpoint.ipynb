```python
fLex4="Lexique4/Lexique4.tsv"
dfLex4=pd.read_csv(fLex4,sep="\t")
dfLex4
```

# Tests sur Lexique4 complet

```python
phon2Lex={}
for d in dfLex4[["3_Phono_IPA","4_Lemme"]].to_dict(orient="records"):
    k=d["3_Phono_IPA"]
    v=d["4_Lemme"]
    if k not in phon2Lex:
        phon2Lex[k]=set()
    phon2Lex[k].add(v)
phon2Lex
```

```python
words=dfLex4["3_Phono_IPA"].tolist()
words=[str(w) for w in words]
words
```

```python
verbs=dfLex4.loc[dfLex4["5_Cgram"]=="VER"]["3_Phono_IPA"].tolist()
verbs=[str(w) for w in verbs]
verbs
```

```python
# 2. Load the list into Linguistica
# 'morpheme_min_len' and 'affix_min_len' can be adjusted based on word length
dLxa = lxa.from_wordlist(words)

# 3. Run morphological analysis
# Extract signatures (prefix/suffix patterns like NULL-ed-ing-s)
signatures = dLxa.signatures()
# signatures 
```

```python
dfLex4.columns
```

```python
lWords=dLxa.wordlist()
affixs=dLxa.affixes()
stems=dLxa.stems()
```

```python
cAffixs={}
for w,s in dLxa.words_to_signatures().items():
    affs=list(s)[0]
    for aff in affs:
        if aff not in cAffixs:
            cAffixs[aff]=[]
        cAffixs[aff].append(w)
for k,v in sorted(cAffixs.items(),key=lambda item: len(item[1]),reverse=True):
    print(k,len(v))
```
